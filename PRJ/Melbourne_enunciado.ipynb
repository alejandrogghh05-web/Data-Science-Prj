## 2. Viviendas mayores que las vistas por el bosque: regresión múltiple

**Responsables:** Judit y Alejandro

**Problema.** Un Random Forest no extrapola: fuera del rango de valores vistos en el entrenamiento, su predicción se queda plana en el valor del extremo. Si un usuario quiere construir una vivienda mayor (más superficie, más habitaciones) que todas las del dataset, el bosque no sabrá responder bien.

**Planteamiento.** Para esos casos se propone una **regresión lineal múltiple**, que sí continúa la tendencia fuera del rango observado.

**Preguntas a responder:**

- ¿Qué variables del bosque se salen de rango en los casos de interés y cuánto?
- ¿Cómo se decide cuándo una vivienda cae "fuera de lo visto" y hay que cambiar de modelo?
- ¿Qué precisión tiene la regresión múltiple dentro del rango frente al bosque, y qué fiabilidad cabe esperar fuera de él?
- ¿Es razonable que la relación precio-superficie siga siendo lineal para casas muy grandes, o hay saturación?

**Entregable.** Una regla clara de uso (qué modelo se aplica en cada caso) y una advertencia sobre los límites de la extrapolación.
